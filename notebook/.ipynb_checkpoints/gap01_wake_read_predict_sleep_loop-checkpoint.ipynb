## Purpose

Notebooks 01 and 02 evaluated next-step occupancy prediction as a static machine-learning problem: every minute, the model is given a fresh, fully up-to-date feature vector and asked to predict `Occupancy(t+1)`. That is a fair way to compare model accuracy, but it quietly assumes the IoT device is reading every sensor every single minute, regardless of which power state it is in.

That assumption is unrealistic, and it is exactly the gap the supervisor flagged in feedback on the Status Report and Project Plan:

- **"Algorithm unclear"** — the scheduler was described in terms of probability thresholds, but never as a concrete step-by-step operational sequence the device actually executes.
- **"Data continuity problem when node sleeps"** — if the device is genuinely powered down in Light Sleep or Deep Sleep, it cannot be sampling sensors every minute. So what does it use for the lag and delta features the model was trained on, once it wakes up several minutes later?

This notebook addresses both points directly by implementing the actual **Wake → Read → Predict → Sleep (WRPS)** loop the device runs in deployment, with state-dependent check intervals (no silent assumption of continuous sampling), and by comparing two ways of handling the resulting data gaps.

## Algorithm

At each wake event the device executes the following five steps:

1. **Wake** — if coming from Light Sleep or Deep Sleep, pay the fixed wake-up energy cost for that state.
2. **Read** — take one sensor reading (Temperature, Humidity, Light, CO2, HumidityRatio).
3. **Build features** — combine the new reading with whatever the *last actually-observed* reading was. This may be 1 minute old (if the device was Active) or up to 8 minutes old (if it was in Deep Sleep) — see "Data continuity" below for how this is handled.
4. **Predict** — run the trained Logistic Regression model (same model and feature set as Notebook 01/02) to get `P = P(Occupancy(t+1) = 1)`.
5. **Decide and sleep** — apply the threshold rule (`P ≥ 0.6 → Active`, `0.3 ≤ P < 0.6 → Light Sleep`, `P < 0.3 → Deep Sleep`) and sleep for that state's check interval before repeating the loop.

Check intervals are a deliberate design parameter, not the dataset's native 1-minute resolution: Active checks every minute (the device is awake anyway), Light Sleep checks every 3 minutes, and Deep Sleep checks every 8 minutes. The longer the sleep, the bigger the data gap the next wake-up has to deal with — this is where the continuity problem actually bites.

```python
import warnings
warnings.filterwarnings('ignore')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
matplotlib.use('Agg')

from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, roc_auc_score

print('Libraries loaded successfully.')
```

## Section 1 — Load data and train the baseline model

This re-trains the same Logistic Regression model from Notebook 01 (identical feature set, identical preprocessing), so this notebook is self-contained and doesn't depend on a pickled model file from an earlier run.

```python
train_path = '../data/datatraining.txt'
test_path  = '../data/datatest2.txt'

train_df = pd.read_csv(train_path)
test_df  = pd.read_csv(test_path)
train_df['date'] = pd.to_datetime(train_df['date'])
test_df['date']  = pd.to_datetime(test_df['date'])

def add_features(df):
    df = df.copy()
    df['hour'] = df['date'].dt.hour
    df['minute'] = df['date'].dt.minute
    df['is_work_hour'] = ((df['hour'] >= 8) & (df['hour'] <= 18)).astype(int)
    df['Light_lag1'] = df['Light'].shift(1)
    df['CO2_lag1'] = df['CO2'].shift(1)
    df['Temp_lag1'] = df['Temperature'].shift(1)
    df['Light_delta'] = df['Light'] - df['Light_lag1']
    df['CO2_delta'] = df['CO2'] - df['CO2_lag1']
    df['Temp_delta'] = df['Temperature'] - df['Temp_lag1']
    df['Light_roll3'] = df['Light'].rolling(3).mean()
    df['CO2_roll3'] = df['CO2'].rolling(3).mean()
    df['Occupancy_t1'] = df['Occupancy'].shift(-1)
    df = df.dropna().reset_index(drop=True)
    df['Occupancy_t1'] = df['Occupancy_t1'].astype(int)
    return df

FEATURE_COLS = ['Temperature', 'Humidity', 'Light', 'CO2', 'HumidityRatio',
                'hour', 'minute', 'is_work_hour',
                'Light_lag1', 'CO2_lag1', 'Temp_lag1',
                'Light_delta', 'CO2_delta', 'Temp_delta',
                'Light_roll3', 'CO2_roll3']
TARGET = 'Occupancy_t1'

train_feat = add_features(train_df)
X_train = train_feat[FEATURE_COLS]
y_train = train_feat[TARGET]

scaler = StandardScaler()
X_train_sc = scaler.fit_transform(X_train)
model = LogisticRegression(max_iter=1000, random_state=42)
model.fit(X_train_sc, y_train)

print('Model trained.')
print(f'Train accuracy: {accuracy_score(y_train, model.predict(X_train_sc)):.4f}')
```

## Section 2 — Operational parameters

Power and energy values are taken from the STM32L476 datasheet assumptions used in the energy modelling work: 10 mW Active, 1 mW Light Sleep, 0.01 mW Deep Sleep, plus small fixed costs for running inference and waking from sleep. The check intervals are the new addition this notebook introduces — they are what make the energy savings and the data-continuity problem both real, instead of assumed away.

```python
P_ACTIVE_mW       = 10.00
P_LIGHT_SLEEP_mW  = 1.00
P_DEEP_SLEEP_mW   = 0.010
E_INFERENCE_mJ    = 0.50
E_WAKEUP_LIGHT_mJ = 0.05
E_WAKEUP_DEEP_mJ  = 0.20

# How many minutes the device sleeps before its next wake-and-check cycle.
# ACTIVE checks every minute (it's awake anyway); the sleep states check
# less often, which is what actually saves energy -- and what creates the
# data gap the next prediction has to deal with.
CHECK_INTERVAL_MIN = {'ACTIVE': 1, 'LIGHT_SLEEP': 3, 'DEEP_SLEEP': 8}
POWER_mW  = {'ACTIVE': P_ACTIVE_mW, 'LIGHT_SLEEP': P_LIGHT_SLEEP_mW, 'DEEP_SLEEP': P_DEEP_SLEEP_mW}
WAKEUP_mJ = {'ACTIVE': 0.0, 'LIGHT_SLEEP': E_WAKEUP_LIGHT_mJ, 'DEEP_SLEEP': E_WAKEUP_DEEP_mJ}

THRESH_HIGH = 0.6
THRESH_LOW  = 0.3

def decide_state(p):
    if p >= THRESH_HIGH:
        return 'ACTIVE'
    elif p >= THRESH_LOW:
        return 'LIGHT_SLEEP'
    else:
        return 'DEEP_SLEEP'

print('Operational parameters set.')
```

## Section 3 — Addressing the data continuity problem

When the device wakes up after several minutes asleep, the lag and delta features it was trained on (`Light_lag1`, `CO2_delta`, etc.) can no longer mean "the value 1 minute ago" — there simply isn't a reading from 1 minute ago. Two strategies are compared:

- **Naive** — use whatever the last actually-observed reading was, but compute the delta as if it were exactly 1 minute old regardless of the true gap. This is the easy-but-wrong approach: a real change accumulated over an 8-minute Deep Sleep gets reported as if it happened in a single minute, inflating the apparent rate of change.
- **Gap-aware** — use the same last-observed reading, but normalise the delta by the true elapsed gap (`Δvalue / gap_minutes`), so the feature still represents a genuine per-minute rate of change.

Both strategies use the same trained model from Section 1 — only the way the feature vector is constructed at prediction time differs.

```python
def build_feature_row(current, buf, gap_minutes, hour, minute, mode):
    """
    current: dict of this minute's true sensor reading
    buf: list of up to the last 3 actually-observed readings (most recent last)
    gap_minutes: minutes elapsed since the previous actual reading (buf[-1])
    mode: 'naive' or 'gapaware'
    """
    is_work_hour = int(8 <= hour <= 18)

    if len(buf) == 0:
        # Cold start: no history yet, fall back to the current reading.
        light_lag1 = current['Light']; co2_lag1 = current['CO2']; temp_lag1 = current['Temperature']
        light_roll3 = current['Light']; co2_roll3 = current['CO2']
    else:
        prev = buf[-1]
        light_lag1, co2_lag1, temp_lag1 = prev['Light'], prev['CO2'], prev['Temperature']
        recent = buf[-3:] + [current]
        light_roll3 = np.mean([r['Light'] for r in recent])
        co2_roll3   = np.mean([r['CO2'] for r in recent])

    if mode == 'naive':
        # Pretends the gap was always 1 minute -- breaks down after long sleeps.
        light_delta = current['Light'] - light_lag1
        co2_delta   = current['CO2']   - co2_lag1
        temp_delta  = current['Temperature'] - temp_lag1
    else:
        # Normalise by the true elapsed gap so this is still a per-minute rate.
        g = max(gap_minutes, 1)
        light_delta = (current['Light'] - light_lag1) / g
        co2_delta   = (current['CO2']   - co2_lag1) / g
        temp_delta  = (current['Temperature'] - temp_lag1) / g

    return [
        current['Temperature'], current['Humidity'], current['Light'], current['CO2'], current['HumidityRatio'],
        hour, minute, is_work_hour,
        light_lag1, co2_lag1, temp_lag1,
        light_delta, co2_delta, temp_delta,
        light_roll3, co2_roll3
    ]

print('Feature builder ready.')
```

## Section 4 — The Wake-Read-Predict-Sleep loop

This is a direct implementation of the five-step algorithm described above. Two things are tracked alongside the energy total: `missed_active_minutes` counts every minute the room was genuinely occupied while the device was asleep and not checking, which is the real-world cost of the data continuity problem — it isn't just a feature-engineering inconvenience, it is missed events.

```python
def run_wrps_loop(df, mode):
    L = len(df)
    i = 0
    state = 'ACTIVE'
    buf = []  # history of actually-observed readings
    last_reading_idx = None
    log = []
    total_energy_mJ = 0.0
    missed_active_minutes = 0
    total_active_minutes_truth = int(df['Occupancy'].sum())

    while i < L:
        row = df.iloc[i]
        hour, minute = row['date'].hour, row['date'].minute
        current = {'Temperature': row['Temperature'], 'Humidity': row['Humidity'],
                   'Light': row['Light'], 'CO2': row['CO2'], 'HumidityRatio': row['HumidityRatio']}

        # 1. WAKE -- pay the wake-up cost if coming from a sleep state
        total_energy_mJ += WAKEUP_mJ[state]

        gap_minutes = 1 if last_reading_idx is None else (i - last_reading_idx)

        # 2-3. READ + BUILD FEATURES (naive vs gap-aware handling of the gap)
        feat = build_feature_row(current, buf, gap_minutes, hour, minute, mode)
        feat_df = pd.DataFrame([feat], columns=FEATURE_COLS)
        feat_sc = scaler.transform(feat_df)

        # 4. PREDICT
        total_energy_mJ += E_INFERENCE_mJ
        p = model.predict_proba(feat_sc)[0, 1]

        # 5. DECIDE next state and sleep for its check interval
        next_state = decide_state(p)
        interval = CHECK_INTERVAL_MIN[next_state]

        if next_state == 'ACTIVE':
            total_energy_mJ += P_ACTIVE_mW * 60  # one full active minute
        else:
            sleep_minutes = interval - 1
            total_energy_mJ += POWER_mW[next_state] * sleep_minutes * 60

        # Evaluation only (the device itself never sees these skipped minutes):
        # did we miss any genuinely-occupied minutes while asleep?
        for j in range(i + 1, min(i + interval, L)):
            if next_state != 'ACTIVE' and df.iloc[j]['Occupancy'] == 1:
                missed_active_minutes += 1

        log.append({'idx': i, 'time': row['date'], 'state_entered': next_state,
                    'P': p, 'gap_minutes': gap_minutes})

        buf.append(current)
        if len(buf) > 3:
            buf.pop(0)
        last_reading_idx = i
        state = next_state
        i += interval

    energy_summary = {
        'mode': mode,
        'total_energy_mJ': total_energy_mJ,
        'missed_active_minutes': missed_active_minutes,
        'total_active_minutes_truth': total_active_minutes_truth,
        'miss_rate_pct': 100 * missed_active_minutes / max(total_active_minutes_truth, 1),
        'n_cycles': len(log),
    }
    return pd.DataFrame(log), energy_summary

log_naive, summary_naive = run_wrps_loop(test_df, mode='naive')
log_gap, summary_gap = run_wrps_loop(test_df, mode='gapaware')

always_active_energy_mJ = P_ACTIVE_mW * 60 * len(test_df)

print('Always-active baseline energy (mJ):', round(always_active_energy_mJ, 1))
print()
print('NAIVE strategy:')
for k, v in summary_naive.items():
    print(f'  {k}: {v}')
print(f"  Energy saving vs always-active: {100 * (1 - summary_naive['total_energy_mJ'] / always_active_energy_mJ):.1f}%")
print()
print('GAP-AWARE strategy:')
for k, v in summary_gap.items():
    print(f'  {k}: {v}')
print(f"  Energy saving vs always-active: {100 * (1 - summary_gap['total_energy_mJ'] / always_active_energy_mJ):.1f}%")
```

## Section 5 — How big is the data gap, really?

A quick sanity check on the gap distribution shows just how far this is from the "fresh data every minute" assumption Notebooks 01 and 02 relied on.

```python
print('Distribution of gap_minutes at each wake-up (naive run):')
print(log_naive['gap_minutes'].describe())
print()
print('Fraction of wake-ups with a gap > 1 minute:',
      f"{(log_naive['gap_minutes'] > 1).mean():.1%}")
```

## Section 6 — Visualising the algorithm in action

The timeline below plots the device's actual state minute-by-minute against ground-truth occupancy for a representative 400-minute stretch of the test set. Shaded red regions are minutes the room was genuinely occupied; the blue step line is the state the device was actually in. Wherever the blue line sits below the red shading, the device was asleep and not checking during a real occupancy event — a visible instance of the data continuity problem.

```python
def expand_state_timeline(log_df, L):
    timeline = np.empty(L, dtype=object)
    for _, r in log_df.iterrows():
        start = r['idx']
        interval = CHECK_INTERVAL_MIN[r['state_entered']]
        end = min(start + interval, L)
        timeline[start:end] = r['state_entered']
    return timeline

STATE_NUM = {'DEEP_SLEEP': 0, 'LIGHT_SLEEP': 1, 'ACTIVE': 2}

tl_naive = expand_state_timeline(log_naive, len(test_df))
tl_gap = expand_state_timeline(log_gap, len(test_df))

window = slice(2700, 3100)  # a 400-minute stretch with active wake/sleep cycling
truth_window = test_df['Occupancy'].values[window]
naive_window = [STATE_NUM[s] for s in tl_naive[window]]
gap_window = [STATE_NUM[s] for s in tl_gap[window]]

fig, axes = plt.subplots(2, 1, figsize=(12, 6), sharex=True)
for ax, states, label in zip(axes, [naive_window, gap_window], ['Naive', 'Gap-aware']):
    x = range(window.start, window.start + len(states))
    ax.step(x, states, where='post', color='#1F77B4', linewidth=1.5, label=f'{label} device state')
    ax.fill_between(x, 0, truth_window * 2.2, step='post', color='#D62728', alpha=0.25, label='Ground-truth occupied')
    ax.set_yticks([0, 1, 2])
    ax.set_yticklabels(['Deep Sleep', 'Light Sleep', 'Active'])
    ax.set_title(f'{label} strategy \u2014 device state vs true occupancy (minutes {window.start}-{window.stop})')
    ax.legend(loc='upper right', fontsize=8)
axes[-1].set_xlabel('Minutes into test set')
plt.tight_layout()
plt.savefig('../figures/wrps_timeline_comparison.png', dpi=150)
plt.show()
print('Saved: ../figures/wrps_timeline_comparison.png')
```

## Section 7 — Energy savings vs missed-activity trade-off

The headline energy-saving figures from earlier notebooks still hold once realistic, state-dependent check intervals are introduced — but they now come with an explicit, measured cost in missed occupancy events, rather than an implicit assumption that data continuity isn't a problem.

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))
labels = ['Always-Active', 'Naive', 'Gap-aware']
energies = [always_active_energy_mJ, summary_naive['total_energy_mJ'], summary_gap['total_energy_mJ']]
miss_rates = [0, summary_naive['miss_rate_pct'], summary_gap['miss_rate_pct']]

axes[0].bar(labels, [e / 1000 for e in energies], color=['#7F7F7F', '#1F77B4', '#2CA02C'])
axes[0].set_ylabel('Total energy (J)')
axes[0].set_title('Energy consumption over the test horizon')

axes[1].bar(labels, miss_rates, color=['#7F7F7F', '#1F77B4', '#2CA02C'])
axes[1].set_ylabel('Missed-activity rate (%)')
axes[1].set_title('Fraction of true occupied minutes missed while asleep')

plt.tight_layout()
plt.savefig('../figures/wrps_energy_vs_miss_rate.png', dpi=150)
plt.show()
print('Saved: ../figures/wrps_energy_vs_miss_rate.png')
```

## Summary

This notebook replaces the implicit "fresh data every minute" assumption from Notebooks 01-02 with a realistic, state-dependent Wake-Read-Predict-Sleep loop, and directly measures what that costs:

- Both the naive and gap-aware strategies retain the large majority of the energy saving over an always-active baseline, confirming the core result still holds under realistic sampling.
- The gap-aware delta normalisation saves slightly more energy than the naive approach in this run, but at the cost of a higher missed-activity rate — because normalising the delta by a long Deep Sleep gap makes a real change in conditions look like a slower, calmer trend, which keeps the scheduler in deeper sleep states for longer. The naive approach's tendency to over-react to long-gap deltas means it wakes up more readily, at a small energy cost.
- The gap distribution shows a meaningful fraction of wake-ups (roughly a quarter or more) inherit a sensor reading that is several minutes stale rather than the 1-minute-fresh reading the original model assumed during training — this is the data continuity problem made concrete and quantified, not just acknowledged.

This trade-off between energy savings and missed-event risk is exactly the kind of design decision the supervisor's feedback was asking the project to make explicit, rather than leaving as an unstated assumption.
