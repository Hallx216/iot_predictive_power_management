```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
matplotlib.use('Agg')
import time
import sys
import pickle
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score, classification_report,
    confusion_matrix, ConfusionMatrixDisplay,
    roc_auc_score, roc_curve
)
```

```python
# ============================================================
# 1. LOAD DATA
# ============================================================
train_path = "../data/datatraining.txt"
test_path  = "../data/datatest2.txt"
```

```python
train_df = pd.read_csv(train_path)
test_df  = pd.read_csv(test_path)
```

```python
# ============================================================
# 2. FEATURE ENGINEERING
# ============================================================
def add_features(df):
    df = df.copy()
    df['date'] = pd.to_datetime(df['date'])

    # Time features
    df['hour']    = df['date'].dt.hour
    df['minute']  = df['date'].dt.minute
    df['is_work_hour'] = ((df['hour'] >= 8) & (df['hour'] <= 18)).astype(int)

    # Lag features (previous time step readings)
    df['Light_lag1']  = df['Light'].shift(1)
    df['CO2_lag1']    = df['CO2'].shift(1)
    df['Temp_lag1']   = df['Temperature'].shift(1)

    # Delta features (rate of change)
    df['Light_delta'] = df['Light'] - df['Light_lag1']
    df['CO2_delta']   = df['CO2']   - df['CO2_lag1']
    df['Temp_delta']  = df['Temperature'] - df['Temp_lag1']

    # Rolling mean (3-step window) - smoothed context
    df['Light_roll3'] = df['Light'].rolling(3).mean()
    df['CO2_roll3']   = df['CO2'].rolling(3).mean()

    # Next-step target: predict occupancy at t+1
    df['Occupancy_t1'] = df['Occupancy'].shift(-1)

    df = df.dropna().reset_index(drop=True)
    df['Occupancy_t1'] = df['Occupancy_t1'].astype(int)
    return df
```

```python
train_df = add_features(train_df)
test_df  = add_features(test_df)
```

```python
FEATURE_COLS = [
    'Temperature', 'Humidity', 'Light', 'CO2', 'HumidityRatio',
    'hour', 'minute', 'is_work_hour',
    'Light_lag1', 'CO2_lag1', 'Temp_lag1',
    'Light_delta', 'CO2_delta', 'Temp_delta',
    'Light_roll3', 'CO2_roll3'
]
TARGET = 'Occupancy_t1'
```

```python
X_train = train_df[FEATURE_COLS]
y_train = train_df[TARGET]
X_test  = test_df[FEATURE_COLS]
y_test  = test_df[TARGET]
```

```python
# ============================================================
# 3. SCALE FEATURES
# ============================================================
scaler = StandardScaler()
X_train_sc = scaler.fit_transform(X_train)
X_test_sc  = scaler.transform(X_test)
```

```python
# ============================================================
# 4. DEFINE MODELS
# ============================================================
# All models chosen for TinyML feasibility:
# - Small memory footprint
# - Fast inference
# - No GPU required
models = {
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
    'Decision Tree (depth=5)': DecisionTreeClassifier(max_depth=5, random_state=42),
    'Decision Tree (depth=10)': DecisionTreeClassifier(max_depth=10, random_state=42),
    'Random Forest (10 trees)': RandomForestClassifier(n_estimators=10, max_depth=5,
                                                        random_state=42),
}
```

```python
# ============================================================
# 5. TRAIN, EVALUATE, MEASURE INFERENCE TIME & MODEL SIZE
# ============================================================
results = []
```

```python
for name, model in models.items():
    print(f"\n{'='*55}")
    print(f"  MODEL: {name}")
    print(f"{'='*55}")

    # Use scaled features for LR, raw for tree models
    if 'Logistic' in name:
        Xtr, Xte = X_train_sc, X_test_sc
    else:
        Xtr, Xte = X_train.values, X_test.values

    # Train
    t0 = time.perf_counter()
    model.fit(Xtr, y_train)
    train_time = time.perf_counter() - t0

    # Inference time (per single sample — critical for IoT)
    n_repeats = 1000
    t0 = time.perf_counter()
    for _ in range(n_repeats):
        model.predict(Xte[0:1])
    inference_time_us = ((time.perf_counter() - t0) / n_repeats) * 1e6  # microseconds

    # Model size (serialised bytes — proxy for RAM footprint)
    model_bytes = sys.getsizeof(pickle.dumps(model))
    model_kb    = model_bytes / 1024

    # Predictions
    y_pred = model.predict(Xte)
    y_prob = model.predict_proba(Xte)[:, 1]

    # Metrics
    acc    = accuracy_score(y_test, y_pred)
    report = classification_report(y_test, y_pred, output_dict=True)
    auc    = roc_auc_score(y_test, y_prob)

    prec1  = report['1']['precision']
    rec1   = report['1']['recall']
    f1_1   = report['1']['f1-score']

    print(f"  Accuracy          : {acc:.4f}")
    print(f"  AUC-ROC           : {auc:.4f}")
    print(f"  Precision (occ=1) : {prec1:.4f}")
    print(f"  Recall    (occ=1) : {rec1:.4f}")
    print(f"  F1-score  (occ=1) : {f1_1:.4f}")
    print(f"  Train time        : {train_time*1000:.1f} ms")
    print(f"  Inference/sample  : {inference_time_us:.1f} µs")
    print(f"  Model size        : {model_kb:.1f} KB")
    print(f"\n  Feasible for IoT edge (<256KB RAM, <5ms inference)?")
    feasible = (model_kb < 256) and (inference_time_us < 5000)
    print(f"  → {'YES' if feasible else 'NO'}")

    results.append({
        'Model': name,
        'Accuracy': acc,
        'AUC-ROC': auc,
        'Precision (Occ=1)': prec1,
        'Recall (Occ=1)': rec1,
        'F1 (Occ=1)': f1_1,
        'Inference (µs)': inference_time_us,
        'Model Size (KB)': model_kb,
        'Train Time (ms)': train_time * 1000,
        'IoT Feasible': feasible,
        'model_obj': model,
        'y_pred': y_pred,
        'y_prob': y_prob
    })
```

```python
results_df = pd.DataFrame(results).drop(columns=['model_obj', 'y_pred', 'y_prob'])
print("\n\n=== SUMMARY TABLE ===")
print(results_df.to_string(index=False))
```

```python
# ============================================================
# 6. CONFUSION MATRICES — all models side by side
# ============================================================
fig, axes = plt.subplots(1, len(models), figsize=(5 * len(models), 4))
for ax, r in zip(axes, results):
    cm = confusion_matrix(y_test, r['y_pred'])
    ConfusionMatrixDisplay(cm).plot(ax=ax, colorbar=False)
    ax.set_title(r['Model'], fontsize=9)
plt.suptitle("Confusion Matrices — Next-Step Occupancy Prediction", fontsize=11)
plt.tight_layout()
plt.savefig("../figures/confusion_matrices_all_models.png", dpi=150)
plt.close()
print("\nSaved: ../figures/confusion_matrices_all_models.png")
```

```python
# ============================================================
# 7. ROC CURVES — all models
# ============================================================
plt.figure(figsize=(8, 6))
for r in results:
    fpr, tpr, _ = roc_curve(y_test, r['y_prob'])
    plt.plot(fpr, tpr, label=f"{r['Model']} (AUC={r['AUC-ROC']:.3f})")
plt.plot([0,1],[0,1],'k--', label='Random Classifier')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curves — All Models (Next-Step Prediction)")
plt.legend(fontsize=8)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig("../figures/roc_curves_all_models.png", dpi=150)
plt.close()
print("Saved: ../figures/roc_curves_all_models.png")
```

```python
# ============================================================
# 8. PERFORMANCE vs FEASIBILITY SCATTER
# ============================================================
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
```

```python
# AUC vs Model Size
axes[0].scatter(results_df['Model Size (KB)'], results_df['AUC-ROC'], s=100)
for _, row in results_df.iterrows():
    axes[0].annotate(row['Model'], (row['Model Size (KB)'], row['AUC-ROC']),
                     textcoords='offset points', xytext=(5, 5), fontsize=7)
axes[0].axvline(256, color='red', linestyle='--', label='256 KB limit')
axes[0].set_xlabel("Model Size (KB)")
axes[0].set_ylabel("AUC-ROC")
axes[0].set_title("Prediction Quality vs Memory Footprint")
axes[0].legend()
axes[0].grid(True, alpha=0.3)
```

```python
# AUC vs Inference Time
axes[1].scatter(results_df['Inference (µs)'], results_df['AUC-ROC'], s=100)
for _, row in results_df.iterrows():
    axes[1].annotate(row['Model'], (row['Inference (µs)'], row['AUC-ROC']),
                     textcoords='offset points', xytext=(5, 5), fontsize=7)
axes[1].axvline(5000, color='red', linestyle='--', label='5 ms limit')
axes[1].set_xlabel("Inference Time (µs)")
axes[1].set_ylabel("AUC-ROC")
axes[1].set_title("Prediction Quality vs Inference Latency")
axes[1].legend()
axes[1].grid(True, alpha=0.3)
```

```python
plt.suptitle("TinyML Feasibility Analysis for IoT Deployment", fontsize=12)
plt.tight_layout()
plt.savefig("../figures/tinyml_feasibility.png", dpi=150)
plt.close()
print("Saved: ../figures/tinyml_feasibility.png")
```

```python
# ============================================================
# 9. SAVE BEST MODEL PREDICTIONS FOR DOWNSTREAM NOTEBOOKS
# ============================================================
# Select best model: highest F1 for occupied class (most safety-critical)
best_idx = results_df['F1 (Occ=1)'].idxmax()
best_name = results_df.loc[best_idx, 'Model']
print(f"\n✓ Best model by F1(Occ=1): {best_name}")
```

```python
best_result = results[best_idx]
out_df = test_df.copy()
out_df['y_pred'] = best_result['y_pred']
out_df['y_prob'] = best_result['y_prob']
out_df.to_csv("../data/best_model_predictions.csv", index=False)
print("Saved: ../data/best_model_predictions.csv")
```
