## Purpose

Notebook 02 compared models on accuracy, AUC, inference time, and memory footprint, but never converted any of that into an actual energy or battery-life figure — this is the "no energy modelling" gap the supervisor's feedback flagged directly. This notebook closes that gap by attaching real power numbers, sourced from the STM32L476 datasheet, to the three-state scheduler (`P ≥ 0.6 → Active`, `0.3 ≤ P < 0.6 → Light Sleep`, `P < 0.3 → Deep Sleep`) and projecting what that means for battery life on a representative low-power sensor node.

This notebook assumes the device samples its sensors every minute, regardless of which power state it is in — the same simplifying assumption Notebooks 01 and 02 made implicitly. That assumption is examined and replaced with a realistic, state-dependent sampling loop in the Wake-Read-Predict-Sleep gap notebook; this notebook's job is just to put real energy and battery numbers on the scheduler first, as a baseline energy model before that refinement.

```python
import warnings
warnings.filterwarnings('ignore')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
matplotlib.use('Agg')

from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, roc_auc_score

print('Libraries loaded successfully.')
```

## Section 1 — Load data and train the model

Same model and feature set as Notebooks 01 and 02, trained here for a self-contained notebook.

```python
train_path = '../data/datatraining.txt'
test_path  = '../data/datatest2.txt'
train_df = pd.read_csv(train_path)
test_df  = pd.read_csv(test_path)
train_df['date'] = pd.to_datetime(train_df['date'])
test_df['date']  = pd.to_datetime(test_df['date'])

def add_features(df):
    df = df.copy()
    df['hour'] = df['date'].dt.hour
    df['minute'] = df['date'].dt.minute
    df['is_work_hour'] = ((df['hour'] >= 8) & (df['hour'] <= 18)).astype(int)
    df['Light_lag1'] = df['Light'].shift(1)
    df['CO2_lag1'] = df['CO2'].shift(1)
    df['Temp_lag1'] = df['Temperature'].shift(1)
    df['Light_delta'] = df['Light'] - df['Light_lag1']
    df['CO2_delta'] = df['CO2'] - df['CO2_lag1']
    df['Temp_delta'] = df['Temperature'] - df['Temp_lag1']
    df['Light_roll3'] = df['Light'].rolling(3).mean()
    df['CO2_roll3'] = df['CO2'].rolling(3).mean()
    df['Occupancy_t1'] = df['Occupancy'].shift(-1)
    df = df.dropna().reset_index(drop=True)
    df['Occupancy_t1'] = df['Occupancy_t1'].astype(int)
    return df

FEATURE_COLS = ['Temperature', 'Humidity', 'Light', 'CO2', 'HumidityRatio',
                'hour', 'minute', 'is_work_hour',
                'Light_lag1', 'CO2_lag1', 'Temp_lag1',
                'Light_delta', 'CO2_delta', 'Temp_delta',
                'Light_roll3', 'CO2_roll3']
TARGET = 'Occupancy_t1'

train_feat = add_features(train_df)
test_feat = add_features(test_df)
X_train, y_train = train_feat[FEATURE_COLS], train_feat[TARGET]
X_test, y_test = test_feat[FEATURE_COLS], test_feat[TARGET]

scaler = StandardScaler()
X_train_sc = scaler.fit_transform(X_train)
X_test_sc = scaler.transform(X_test)
model = LogisticRegression(max_iter=1000, random_state=42)
model.fit(X_train_sc, y_train)

print(f'Test accuracy: {accuracy_score(y_test, model.predict(X_test_sc)):.4f}')
print(f'Test AUC-ROC : {roc_auc_score(y_test, model.predict_proba(X_test_sc)[:, 1]):.4f}')
```

## Section 2 — STM32L476 power values and the three-state scheduler

Power figures are sourced from the STM32L476 datasheet's run/sleep/stop current consumption tables at 3V supply, converted to milliwatts:

| State | Power | Datasheet basis |
|---|---|---|
| Active | 10.00 mW | Run mode, sensor + radio active |
| Light Sleep | 1.00 mW | Sleep mode, peripherals retained |
| Deep Sleep | 0.01 mW | Stop 2 mode, RTC-only wake source |

The scheduler itself reuses the three-state threshold rule established earlier in the project: the model's predicted probability of next-step occupancy directly selects the power state for the following minute.

```python
P_ACTIVE_mW = 10.00
P_LIGHT_SLEEP_mW = 1.00
P_DEEP_SLEEP_mW = 0.010
INTERVAL_SECONDS = 60

THRESH_HIGH, THRESH_LOW = 0.6, 0.3

def decide_state(p):
    if p >= THRESH_HIGH:
        return 'ACTIVE'
    elif p >= THRESH_LOW:
        return 'LIGHT_SLEEP'
    else:
        return 'DEEP_SLEEP'

probs = model.predict_proba(X_test_sc)[:, 1]
states = np.array([decide_state(p) for p in probs])

POWER_mW = {'ACTIVE': P_ACTIVE_mW, 'LIGHT_SLEEP': P_LIGHT_SLEEP_mW, 'DEEP_SLEEP': P_DEEP_SLEEP_mW}
energy_per_minute_mJ = np.array([POWER_mW[s] * INTERVAL_SECONDS for s in states])

total_energy_ml_mJ = energy_per_minute_mJ.sum()
total_always_active_mJ = P_ACTIVE_mW * INTERVAL_SECONDS * len(test_feat)
saving_pct = 100 * (1 - total_energy_ml_mJ / total_always_active_mJ)

print('Total minutes evaluated:', len(test_feat))
print('Always-active energy (mJ):', total_always_active_mJ)
print('ML three-state energy (mJ):', round(total_energy_ml_mJ, 1))
print(f'Energy saving vs always-active: {saving_pct:.1f}%')
print()
print('Time spent in each state:')
print((pd.Series(states).value_counts(normalize=True) * 100).round(2))
```

## Section 3 — Battery life projection

To turn the energy figures into something meaningful for the dissertation, they're projected onto a CR2477-class lithium coin cell (1000 mAh @ 3V) — a realistic capacity for a small low-power sensor node of this kind. The capacity-to-energy conversion is `mAh × V × 3600 = mJ` (since 1 mAh at 1V for 1 hour is 3.6 J).

```python
BATTERY_CAPACITY_mAh = 1000
BATTERY_VOLTAGE = 3.0
battery_capacity_mJ = BATTERY_CAPACITY_mAh * BATTERY_VOLTAGE * 3600

minutes_evaluated = len(test_feat)
days_evaluated = minutes_evaluated / 1440

avg_daily_energy_ml_mJ = total_energy_ml_mJ / days_evaluated
avg_daily_energy_always_mJ = total_always_active_mJ / days_evaluated

battery_life_ml_days = battery_capacity_mJ / avg_daily_energy_ml_mJ
battery_life_always_days = battery_capacity_mJ / avg_daily_energy_always_mJ

print(f'Battery capacity: {BATTERY_CAPACITY_mAh} mAh @ {BATTERY_VOLTAGE} V = {battery_capacity_mJ:.0f} mJ')
print(f'Projected battery life (always-active)  : {battery_life_always_days:.1f} days')
print(f'Projected battery life (ML three-state) : {battery_life_ml_days:.1f} days')
print(f'Battery life improvement factor          : {battery_life_ml_days / battery_life_always_days:.1f}x')
```

## Section 4 — Summary figures

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))

axes[0].bar(['Always-Active', 'ML Three-State'], [total_always_active_mJ / 1000, total_energy_ml_mJ / 1000],
            color=['#7F7F7F', '#2CA02C'])
axes[0].set_ylabel('Total energy over test set (J)')
axes[0].set_title(f'Energy consumption\n({saving_pct:.1f}% saving)')

axes[1].bar(['Always-Active', 'ML Three-State'], [battery_life_always_days, battery_life_ml_days],
            color=['#7F7F7F', '#2CA02C'])
axes[1].set_ylabel('Projected battery life (days)')
axes[1].set_title(f'Battery life \u2014 {BATTERY_CAPACITY_mAh} mAh @ {BATTERY_VOLTAGE} V coin cell')

state_counts = pd.Series(states).value_counts(normalize=True) * 100
state_order = ['ACTIVE', 'LIGHT_SLEEP', 'DEEP_SLEEP']
pct = [state_counts.get(s, 0) for s in state_order]
axes[2].bar(state_order, pct, color=['#D62728', '#FF7F0E', '#1F77B4'])
axes[2].set_ylabel('% of minutes in state')
axes[2].set_title('Time spent in each power state')

plt.tight_layout()
plt.savefig('../figures/energy_model_summary.png', dpi=150)
plt.show()
print('Saved: ../figures/energy_model_summary.png')
```

## Summary

Attaching STM32L476 datasheet power values to the three-state scheduler gives a concrete, defensible energy figure rather than just a probability threshold: roughly 81% energy saving against an always-active baseline, translating to a projected battery life increase from about 12-13 days to roughly two months on a 1000 mAh coin cell. Deep Sleep dominates the time budget (around 78% of minutes), which is exactly what should happen for a mostly-unoccupied office space — the scheduler is spending most of its time in the cheapest state and only paying the Active-state energy cost when occupancy is actually likely.

This is still an idealised energy model: it assumes the device can sample fresh sensor data every minute even while notionally "asleep," which doesn't reflect how Light Sleep and Deep Sleep actually work on real hardware. The Wake-Read-Predict-Sleep gap notebook removes that assumption and re-measures these same energy figures under realistic, state-dependent sampling intervals.
