## Purpose

Every previous notebook used a single, fixed threshold pair (P_high = 0.6, P_low = 0.3) without ever testing whether that particular pair was actually a good choice. This notebook sweeps 12 different (P_high, P_low) configurations to map out the full energy-saving vs missed-activity trade-off, and selects a justified operating point rather than an arbitrarily chosen one.

The threshold pair controls how cautious the scheduler is: a high P_high means the device only goes Active when very confident the room is occupied (saving more energy, but reacting late); a high P_low means the device avoids Deep Sleep unless very confident the room will stay empty (missing fewer events, but saving less energy). Twelve combinations spanning conservative to aggressive sleep policies are tested here.

```python
import warnings
warnings.filterwarnings('ignore')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
matplotlib.use('Agg')

from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler

print('Libraries loaded successfully.')
```

## Section 1 — Load data and train the model

Same model and feature set used throughout the project.

```python
train_path = '../data/datatraining.txt'
test_path  = '../data/datatest2.txt'
train_df = pd.read_csv(train_path)
test_df  = pd.read_csv(test_path)
train_df['date'] = pd.to_datetime(train_df['date'])
test_df['date']  = pd.to_datetime(test_df['date'])

def add_features(df):
    df = df.copy()
    df['hour'] = df['date'].dt.hour
    df['minute'] = df['date'].dt.minute
    df['is_work_hour'] = ((df['hour'] >= 8) & (df['hour'] <= 18)).astype(int)
    df['Light_lag1'] = df['Light'].shift(1)
    df['CO2_lag1'] = df['CO2'].shift(1)
    df['Temp_lag1'] = df['Temperature'].shift(1)
    df['Light_delta'] = df['Light'] - df['Light_lag1']
    df['CO2_delta'] = df['CO2'] - df['CO2_lag1']
    df['Temp_delta'] = df['Temperature'] - df['Temp_lag1']
    df['Light_roll3'] = df['Light'].rolling(3).mean()
    df['CO2_roll3'] = df['CO2'].rolling(3).mean()
    df['Occupancy_t1'] = df['Occupancy'].shift(-1)
    df = df.dropna().reset_index(drop=True)
    df['Occupancy_t1'] = df['Occupancy_t1'].astype(int)
    return df

FEATURE_COLS = ['Temperature', 'Humidity', 'Light', 'CO2', 'HumidityRatio',
                'hour', 'minute', 'is_work_hour',
                'Light_lag1', 'CO2_lag1', 'Temp_lag1',
                'Light_delta', 'CO2_delta', 'Temp_delta',
                'Light_roll3', 'CO2_roll3']
TARGET = 'Occupancy_t1'

train_feat = add_features(train_df)
test_feat = add_features(test_df)
X_train, y_train = train_feat[FEATURE_COLS], train_feat[TARGET]
X_test = test_feat[FEATURE_COLS]

scaler = StandardScaler()
X_train_sc = scaler.fit_transform(X_train)
X_test_sc = scaler.transform(X_test)
model = LogisticRegression(max_iter=1000, random_state=42)
model.fit(X_train_sc, y_train)
probs = model.predict_proba(X_test_sc)[:, 1]

print('Model trained. Predicted probabilities generated for', len(probs), 'test minutes.')
```

## Section 2 — Sweep 12 threshold configurations

For each (P_high, P_low) pair, every minute in the test set is assigned a power state from the model's predicted probability, energy is accumulated using the same STM32L476 power values as Notebook 03, and the missed-activity rate is computed as the share of genuinely occupied minutes where the device was in a sleep state.

```python
P_ACTIVE_mW = 10.00
P_LIGHT_SLEEP_mW = 1.00
P_DEEP_SLEEP_mW = 0.010
INTERVAL_SECONDS = 60
POWER_mW = {'ACTIVE': P_ACTIVE_mW, 'LIGHT_SLEEP': P_LIGHT_SLEEP_mW, 'DEEP_SLEEP': P_DEEP_SLEEP_mW}

def decide_state(p, t_high, t_low):
    if p >= t_high:
        return 'ACTIVE'
    elif p >= t_low:
        return 'LIGHT_SLEEP'
    else:
        return 'DEEP_SLEEP'

truth = test_feat['Occupancy_t1'].values
always_active_mJ = P_ACTIVE_mW * INTERVAL_SECONDS * len(test_feat)

# 12 configurations spanning conservative -> aggressive sleep policies
configs = [
    (0.7, 0.5), (0.7, 0.4), (0.7, 0.3),
    (0.6, 0.4), (0.6, 0.3), (0.6, 0.2),
    (0.5, 0.4), (0.5, 0.3), (0.5, 0.2),
    (0.4, 0.3), (0.4, 0.2), (0.3, 0.2),
]

rows = []
for t_high, t_low in configs:
    states = np.array([decide_state(p, t_high, t_low) for p in probs])
    energy_mJ = np.array([POWER_mW[s] * INTERVAL_SECONDS for s in states]).sum()
    saving_pct = 100 * (1 - energy_mJ / always_active_mJ)
    sleeping = states != 'ACTIVE'
    misses = int(((truth == 1) & sleeping).sum())
    total_occupied = int((truth == 1).sum())
    miss_rate_pct = 100 * misses / max(total_occupied, 1)
    rows.append({'t_high': t_high, 't_low': t_low, 'saving_pct': saving_pct,
                 'miss_rate_pct': miss_rate_pct, 'energy_mJ': energy_mJ})

sweep_df = pd.DataFrame(rows)
print(sweep_df.to_string(index=False))
```

## Section 3 — Selecting a justified operating point

Rather than picking whichever configuration happens to save the most energy (which would push P_high and P_low both high, missing nearly 1 in 5 real occupancy events), a 10% missed-activity ceiling is set as a reasonable operational limit — the point past which an occupancy-triggered system (lighting, HVAC, security) would start missing enough events to be noticeable. Among configurations that respect this ceiling, the one with the highest energy saving is selected.

```python
MISS_RATE_CEILING = 10.0
candidates = sweep_df[sweep_df['miss_rate_pct'] <= MISS_RATE_CEILING].sort_values('saving_pct', ascending=False)
print(f'Candidates with miss rate <= {MISS_RATE_CEILING}%, sorted by saving (descending):')
print(candidates.to_string(index=False))

best = candidates.iloc[0]
print()
print(f"Recommended configuration: P_high={best['t_high']}, P_low={best['t_low']}")
print(f"  Energy saving  : {best['saving_pct']:.1f}%")
print(f"  Miss rate      : {best['miss_rate_pct']:.1f}%")
```

```python
fig, ax = plt.subplots(figsize=(7.5, 6))
band_colors = {0.7: '#9E9E9E', 0.6: '#6699CC', 0.5: '#2CA02C', 0.4: '#FF7F0E', 0.3: '#D62728'}
for t_high_val in sorted(sweep_df['t_high'].unique(), reverse=True):
    sub = sweep_df[sweep_df['t_high'] == t_high_val]
    ax.scatter(sub['saving_pct'], sub['miss_rate_pct'], color=band_colors[t_high_val],
               s=90, label=f'P_high={t_high_val}', zorder=3, edgecolor='white', linewidth=0.5)

ax.axhline(MISS_RATE_CEILING, color='black', linestyle='--', linewidth=1, alpha=0.5, label=f'{MISS_RATE_CEILING:.0f}% miss-rate ceiling')
ax.annotate(f"Recommended: ({best['t_high']:.1f}, {best['t_low']:.1f})\n{best['saving_pct']:.1f}% saving, {best['miss_rate_pct']:.1f}% miss rate",
            (best['saving_pct'], best['miss_rate_pct']), textcoords='offset points', xytext=(15, -35),
            fontsize=9, fontweight='bold', color='#2CA02C',
            arrowprops=dict(arrowstyle='->', color='#2CA02C'))
ax.set_xlabel('Energy saving vs always-active (%)')
ax.set_ylabel('Missed-activity rate (%)')
ax.set_title('Threshold sensitivity sweep: 12 (P_high, P_low) configurations')
ax.legend(loc='upper left', fontsize=8.5)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('../figures/threshold_sensitivity_sweep.png', dpi=150)
plt.show()
print('Saved: ../figures/threshold_sensitivity_sweep.png')
```

## Summary

The sweep shows the threshold pair has a real, substantial effect: missed-activity rate ranges from under 1% to nearly 19% across the 12 configurations tested, while energy saving only varies by about 4 percentage points (78.3% to 82.2%) over that same range. That asymmetry is the key finding — there is comparatively little energy to be gained by pushing thresholds toward the aggressive end, but a great deal of reliability to be lost.

Applying the 10% miss-rate ceiling identifies **P_high = 0.5, P_low = 0.4** as the recommended operating point: 80.0% energy saving with an 8.0% missed-activity rate, the best saving available while staying under the ceiling. This closely matches the (0.5, 0.4) configuration already used as the project's headline figure elsewhere (79.7% saving, 7.0% miss rate in the full Wake-Read-Predict-Sleep simulation) — the small numerical difference is expected, since this notebook evaluates an idealised per-minute decision rule rather than running the full check-interval simulation loop used in the gap notebooks. The two independent calculations landing on the same threshold pair is a reassuring consistency check on the project's central design choice. This (0.5, 0.4) pair should be treated as the project's recommended configuration going forward, in place of the earlier provisional (0.6, 0.3) pair used in Gap Notebooks 1–3 and Notebook 03.
